  Cadeias com uma quantidade par de símbolos. {ε, bb, ac, aabc, abac, abbc, abcc, acac, acbc, aaaacb, bababc, ... } Sigma = {a,b,c} S -> XXS X -> a X -> b X -> c S -> ε ((a|b|c)(a|b|c))*



O código implementa um **Autômato Finito Determinístico (AFD)** com o propósito exclusivo de verificar se a cadeia de entrada, composta pelos símbolos *{a,b,c}*, possui comprimento par. Esse autômato opera com dois estados, "`q_par`" e "`q_impar`". A lógica de transição é simples: a leitura de qualquer símbolo (a,b ou c) faz o autômato alterar entre "`q_par`" e "`q_impar`", que no final da cadeia a função "`automato`" retorna TRUE se o estado final "`q_par`", confirmando que a cadeia tem um número par de símbolos, e FALSE caso contrário.

```python
# Autômato Finito para cadeias de comprimento par sobre {a, b, c}

# Alternando se a,b,c são par ou impar, sendo (estado atual, simbolo)
# E o valor do proximo estado
delta = {
    ('q_par', 'a'): 'q_impar',
    ('q_par', 'b'): 'q_impar',
    ('q_par', 'c'): 'q_impar',
    ('q_impar', 'a'): 'q_par',
    ('q_impar', 'b'): 'q_par',
    ('q_impar', 'c'): 'q_par',
}

# Estado inicial do automato
estado_inicial = 'q_par'
# Definindo que o estado de aceitação é par
estados_finais = {'q_par'}

def automato(cadeia: str) -> bool:
    """
    Simula o autômato finito para verificar se a cadeia tem comprimento par.
    Retorna True se for aceita, False caso contrário.
    """
    estado_atual = estado_inicial # Começa com 'q_par'

    #Percorre cada símbolo da cadeia
    for simbolo in cadeia:
        # Verifica se tem transição para (estado_atual, simbolo)
        # E se o símbolo é valido
        if (estado_atual, simbolo) not in delta:
            print(f"Símbolo inválido: {simbolo}")
            return False
        # Atualiza o estado com base na transição
        estado_atual = delta[(estado_atual, simbolo)]

    # Aceita se terminar no parametro do estado final (par)
    return estado_atual in estados_finais


# ------------------- Execução -------------------
if __name__ == "__main__":
    cadeia = input("Digite uma cadeia com símbolos {a,b,c}: ").strip()
    if automato(cadeia):
        print(f"Cadeia '{cadeia}' -> ACEITA")
    else:
        print(f"Cadeia '{cadeia}' -> REJEITADA")
```
